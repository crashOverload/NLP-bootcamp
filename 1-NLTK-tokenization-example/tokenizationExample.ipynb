```python
corpus = """How to reclaim your focus and amplify its effects.
Leaders are under constant pressure to deliver results, navigate relentless change! But let's get real—how much of what fills your calendar is actually making a difference?"""

print(corpus)
```

```python
##  Tokenization
##  Praagraph --> Sentence 
import nltk
from nltk.tokenize import sent_tokenize
##nltk.download('punkt_tab') # needed for the first time
nltk.data.path.append('C:\\Users\\abhiyank.saxena\\AppData\\Roaming\\nltk_data')
documents = sent_tokenize(corpus)
print(documents)
```

```python
from nltk.tokenize import word_tokenize
word_tokenize(corpus)

for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizr = TreebankWordTokenizer()
tokenizr.tokenize(corpus)
```
